```python
import pandas as pd
```

```python
df = pd.read_csv('height_weight.csv')
df.head()
```

```python
df.shape
```

```python
df.height.describe()
```

```python
height_q1 = df.height.quantile(0.25)
height_q3 = df.height.quantile(0.75)
height_q3, height_q1
```

```python
height_iqr = height_q3 - height_q1
height_iqr
```

```python
height_lowerlimit = height_q1 - 1.5 * height_iqr
height_upperlimit = height_q3 + 1.5 * height_iqr
height_lowerlimit, height_upperlimit
```

```python
df_height_outliers = df[(df.height > height_upperlimit) | (df.height < height_lowerlimit)]
```

```python
#for weights
df.weight.describe()
```

```python
weight_q1 = df.weight.quantile(0.25)
weight_q3 = df.weight.quantile(0.75)
weight_q1, weight_q3
```

```python
weight_iqr = weight_q3 - weight_q1
weight_iqr
```

```python
weight_lowerlimit = weight_q1 - 1.5 * weight_iqr
weight_upperlimit = weight_q3 + 1.5 * weight_iqr
weight_lowerlimit, weight_upperlimit
```

```python
df_weight_outliers = df[(df.weight < weight_lowerlimit)|(df.weight > weight_upperlimit)]
df_weight_outliers
```

```python
df_cleaned = df[~df.index.isin(df_weight_outliers.index) & ~df.index.isin(df_height_outliers.index)]
df_cleaned.shape
```
